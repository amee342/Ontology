```python
onto = get_ontology("file://hi_ontology_extended_populated.owl").load()
hi = onto.get_namespace("http://www.semanticweb.org/vbr240/ontologies/2022/4/untitled-ontology-51/")
g = onto.world.as_rdflib_graph()

SKOS = Namespace("http://www.w3.org/2004/02/skos/core#")
```

```python
def ent(name):
    x = getattr(hi, name, None)
    if x is None:
        print(f"[WARN] Entity not found: {name}")
    return x

def add_link(local_name, predicate, external_uri):
    x = ent(local_name)
    if x is not None:
        g.add((URIRef(x.iri), predicate, URIRef(external_uri)))
```
